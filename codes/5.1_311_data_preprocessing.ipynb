```python
CBG_df = pd.read_csv('../Data/CBG_pop_2010/DECENNIALPL2010.P1-Data.csv')
CBG_df = CBG_df.iloc[1:,:]
CBG_df = CBG_df[["GEO_ID","P001001"]]
CBG_df["CBG_ID"] = CBG_df["GEO_ID"].apply(lambda x:x.split("US")[1].strip()[1:])
CBG_df["P001001"] = pd.to_numeric(CBG_df["P001001"], errors='coerce')
grouped_df = CBG_df.groupby("CBG_ID")["P001001"].sum().reset_index()
grouped_df.rename(columns={"P001001": "CBG_pop"}, inplace=True)
grouped_df
```

```python
# # 删掉REQCATEGOR为NAN的行
# OAK311_spjoin_CB_2015_dropna = OAK311_spjoin_CB_2015_dropna.dropna(subset = ['REQCATEGOR'])
# pivot_table = OAK311_spjoin_CB_2015_dropna.pivot_table(index='GEOID10', columns='REQCATEGOR', aggfunc='size', fill_value=0).reset_index()
# pivot_table['sum_count'] = pivot_table.iloc[:,1:].sum(axis=1)

datasets = [OAK311_spjoin_CB_2015, OAK311_spjoin_CB_2016, OAK311_spjoin_CB_2017,
            OAK311_spjoin_CB_2018, OAK311_spjoin_CB_2019, OAK311_spjoin_CB_2020,
            OAK311_spjoin_CB_2021]
# datasets = [OAK311_spjoin_CB_2021]
#print(len(OAK311_spjoin_CB_2021))

pivot_tables = []

# 循环处理每一个数据集
for data in datasets:
    data = data.dropna(subset=['GEOID10'])
    data_dropna = data.dropna(subset=['REQCATEGOR'])
      
    pivot_table = data_dropna.pivot_table(index='GEOID10', columns='REQCATEGOR', aggfunc='size', fill_value=0).reset_index()
    pivot_table['sum_count'] = pivot_table.iloc[:, 1:].sum(axis=1)
    pivot_tables.append(pivot_table)
```

```python
OAK311_categories_count_CB_2015 = pivot_tables[0]
OAK311_categories_count_CB_2016 = pivot_tables[1]
OAK311_categories_count_CB_2017 = pivot_tables[2]
OAK311_categories_count_CB_2018 = pivot_tables[3]
OAK311_categories_count_CB_2019 = pivot_tables[4]
OAK311_categories_count_CB_2020 = pivot_tables[5]
OAK311_categories_count_CB_2021 = pivot_tables[6]
```

```python
OAK311_categories_count_CB_2015
```

```python
OAK311_categories_count_CB_2015["CBG_ID"] = OAK311_categories_count_CB_2015["GEOID10"].apply(lambda x:x[1:12])
OAK311_categories_count_CB_2015_merge = OAK311_categories_count_CB_2015.merge(grouped_df, on = 'CBG_ID', how = 'left')
OAK311_categories_count_CB_2016["CBG_ID"] = OAK311_categories_count_CB_2016["GEOID10"].apply(lambda x:x[1:12])
OAK311_categories_count_CB_2016_merge = OAK311_categories_count_CB_2016.merge(grouped_df, on = 'CBG_ID', how = 'left')
OAK311_categories_count_CB_2017["CBG_ID"] = OAK311_categories_count_CB_2017["GEOID10"].apply(lambda x:x[1:12])
OAK311_categories_count_CB_2017_merge = OAK311_categories_count_CB_2017.merge(grouped_df, on = 'CBG_ID', how = 'left')
OAK311_categories_count_CB_2018["CBG_ID"] = OAK311_categories_count_CB_2018["GEOID10"].apply(lambda x:x[1:12])
OAK311_categories_count_CB_2018_merge = OAK311_categories_count_CB_2018.merge(grouped_df, on = 'CBG_ID', how = 'left')
OAK311_categories_count_CB_2019["CBG_ID"] = OAK311_categories_count_CB_2019["GEOID10"].apply(lambda x:x[1:12])
OAK311_categories_count_CB_2019_merge = OAK311_categories_count_CB_2019.merge(grouped_df, on = 'CBG_ID', how = 'left')
OAK311_categories_count_CB_2020["CBG_ID"] = OAK311_categories_count_CB_2020["GEOID10"].apply(lambda x:x[1:12])
OAK311_categories_count_CB_2020_merge = OAK311_categories_count_CB_2020.merge(grouped_df, on = 'CBG_ID', how = 'left')
OAK311_categories_count_CB_2021["CBG_ID"] = OAK311_categories_count_CB_2021["GEOID10"].apply(lambda x:x[1:12])
OAK311_categories_count_CB_2021_merge = OAK311_categories_count_CB_2021.merge(grouped_df, on = 'CBG_ID', how = 'left')
```
